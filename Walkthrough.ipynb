# Abstract

Momentum investing is one of the most extensively documented anomalies in financial markets. The central idea is that securities which have performed strongly in the recent past often continue to outperform over intermediate investment horizons.

This project investigates whether machine-learning algorithms can improve the extraction of information from momentum signals when forecasting future stock returns. Using daily price data from S&P 500 constituents between January 2018 and January 2025, a set of momentum-based factors is constructed and used to predict one-month forward returns.

Six machine-learning models are evaluated: Linear Regression, Ridge Regression, Lasso Regression, Random Forest, Gradient Boosting and a Multi-Layer Perceptron (MLP) Neural Network. Predicted returns are transformed into monthly stock rankings, from which equal-weight portfolios are constructed.

Although all models produced negative out-of-sample R² values, every model generated positive investment returns. The MLP Neural Network achieved the highest cumulative return (74.98%), while Random Forest achieved the highest Sharpe ratio (1.06) and lowest maximum drawdown (-0.44%). These findings suggest that successful investment models may depend more on ranking securities correctly than on accurately forecasting future returns.

# 1. Introduction

Financial markets have long exhibited patterns that challenge the assumptions of perfectly efficient markets. Among the most persistent and widely studied of these patterns is the momentum effect.

Momentum suggests that assets which have recently outperformed tend to continue outperforming, while recent underperformers often continue to lag. Traditional momentum strategies typically rank securities according to historical returns and allocate capital to recent winners.

While such approaches have shown success historically, they rely on relatively simple decision rules. Machine-learning methods provide a potential alternative by allowing more complex relationships between momentum factors and future returns to be learned directly from data.

This project addresses the following research question:

> Can machine-learning algorithms improve stock selection using momentum factors, and which algorithm produces the strongest investment performance?

# 4. Momentum Factor Construction

The project intentionally adopts a pure momentum framework.

No fundamental variables are used.

No valuation metrics are included.

No accounting information is incorporated.

This isolates the predictive contribution of momentum signals alone.

## Daily Return

$$
R_t = \frac{P_t - P_{t-1}}{P_{t-1}}
$$

This captures very recent price movement and provides information regarding short-term market behaviour.

## Three-Month Momentum

$$
M_{3M} = \frac{P_t - P_{t-63}}{P_{t-63}}
$$

Measures medium-term trend strength over approximately one quarter.

## Six-Month Momentum

$$
M_{6M} = \frac{P_t - P_{t-126}}{P_{t-126}}
$$

Captures longer-term momentum persistence and sustained market trends.

## Nine-Month Momentum

$$
M_{9M} = \frac{P_t - P_{t-189}}{P_{t-189}}
$$

Provides a measure of trend behaviour over a prolonged investment horizon.

## Twelve-Month Momentum

$$
M_{12M} = \frac{P_t - P_{t-252}}{P_{t-252}}
$$

The twelve-month horizon is one of the most widely studied and implemented momentum measures within quantitative finance.

## Why Multiple Momentum Horizons?

Momentum may operate differently across investment horizons.

Short-term trends can be influenced by investor sentiment and behavioural biases, while longer-term trends may reflect institutional capital flows and gradual information diffusion.

By including multiple momentum horizons, the machine-learning models can identify which timescales contain the strongest predictive information.

# 5. Prediction Target

The objective of the project is to predict future one-month stock returns.

The target variable is defined as:

$$
FutureReturn_t = \frac{P_{t+21} - P_t}{P_t}
$$

where approximately twenty-one trading days corresponds to one calendar month.

The target variable is shifted forward using:

```python
.shift(-21)
```

This ensures that future returns are aligned correctly with information available at the prediction date.

Using a forward-shifted target prevents look-ahead bias and more accurately reflects a realistic investment process.

The prediction task can therefore be summarised as:

> Using momentum information available today, estimate which stocks are most likely to generate the strongest returns over the next month.

Importantly, the objective is not necessarily to predict exact future returns. In quantitative investing, correctly ranking securities is often more valuable than generating highly accurate point forecasts.

This distinction becomes particularly important when interpreting the final results, where profitable portfolios were constructed despite relatively weak prediction accuracy metrics.

# 8. Evaluation Metrics

The models are evaluated using four metrics.

### Total Return

Measures cumulative growth in portfolio value.

### Sharpe Ratio

Measures risk-adjusted performance.

Higher Sharpe ratios indicate superior returns relative to risk.

### Maximum Drawdown

Measures the largest peak-to-trough decline experienced by the portfolio.

Lower drawdowns indicate better downside protection.

### R²

Measures predictive accuracy.

Importantly, profitable investment strategies do not necessarily require highly accurate return forecasts. In quantitative investing, ranking securities correctly is often more important than predicting exact future returns.

# 10. Prediction Accuracy Versus Investment Performance

One of the most interesting findings is that every model produced a negative out-of-sample R² value.

In many machine-learning applications, negative R² values would be interpreted as evidence of poor predictive performance.

However, every portfolio generated positive returns, with several significantly outperforming the benchmark linear models.

This apparent contradiction highlights an important principle of quantitative investing:

> A model does not need to accurately predict future returns to be economically useful.

Instead, the model only needs to rank stocks more effectively than random selection.

The results suggest that ranking ability may be more important than forecast accuracy when constructing machine-learning investment strategies.

# 12. Discussion

Several important observations emerge from the results.

First, all six machine-learning models generated positive portfolio returns, suggesting that momentum remained a useful source of information throughout the sample period.

Second, nonlinear models consistently outperformed traditional linear techniques. Both Random Forest and the MLP Neural Network produced materially stronger performance than ordinary regression models.

Third, model complexity does not automatically guarantee superiority. Although the neural network generated the highest return, Random Forest achieved better risk-adjusted performance and significantly lower downside risk.

Finally, the weak relationship between R² and portfolio returns suggests that successful investment strategies depend more heavily on ranking securities than forecasting precise returns.

# 13. Limitations

### Survivorship Bias

The study uses current S&P 500 constituents and therefore excludes firms that left the index during the sample period.

### Transaction Costs

### Hyperparameter Optimisation

The models were evaluated largely using default parameter settings.

### Neural Network Simplicity

The model architecture remains relatively simple and was not extensively tuned.

### Single Train-Test Split

A walk-forward validation framework would provide more robust performance estimates.

### Limited Factor Set

Only momentum factors were included. Additional variables such as value, quality and volatility factors may improve predictive performance.
